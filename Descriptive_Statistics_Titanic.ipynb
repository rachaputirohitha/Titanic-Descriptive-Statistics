```python
import pandas as pd

df = pd.read_csv("/titanic.csv")

df.head()
```

```python
df[["age", "fare", "sibsp", "parch"]].describe()
```

```python
df[["age", "fare", "sibsp", "parch"]].mode()
```

```python
summary = pd.DataFrame({
    "Mean": df[["age", "fare", "sibsp", "parch"]].mean(),
    "Median": df[["age", "fare", "sibsp", "parch"]].median(),
    "Mode": df[["age", "fare", "sibsp", "parch"]].mode().iloc[0],
    "Std Deviation": df[["age", "fare", "sibsp", "parch"]].std(),
    "25th Percentile": df[["age", "fare", "sibsp", "parch"]].quantile(0.25),
    "50th Percentile": df[["age", "fare", "sibsp", "parch"]].quantile(0.50),
    "75th Percentile": df[["age", "fare", "sibsp", "parch"]].quantile(0.75)
})

summary
```

```python
summary.round(2)
```

```python
skewness = df[["age", "fare", "sibsp", "parch"]].skew()

skewness.round(2)
```

```python
q1 = df[["age", "fare", "sibsp", "parch"]].quantile(0.25)
q3 = df[["age", "fare", "sibsp", "parch"]].quantile(0.75)

iqr = q3 - q1

iqr.round(2)
```

Descriptive Statistics Interpretation

Age: The mean age is about 29.70 years, while the median is 28 years. The mean is slightly higher than the median, indicating a small positive skew. The standard deviation of about 14.53 shows noticeable variation in passenger ages. The middle 50% of ages lie between approximately 20.13 and 38 years.

Fare: The mean fare is about 32.20, while the median is 14.45. The much higher mean and positive skewness indicate that the fare distribution is strongly right-skewed. The standard deviation of about 49.69 also shows high variability. The middle 50% of fares lie between approximately 7.91 and 31.

SibSp: The median and mode are both 0, meaning most passengers had no siblings or spouses aboard. The positive skewness indicates that a smaller number of passengers had higher SibSp values. The middle 50% of the data ranges from 0 to 1.

Parch: The median and mode are both 0, showing that most passengers had no parents or children aboard. The positive skewness indicates that a smaller number of passengers had higher Parch values. The IQR is 0 because both the 25th and 75th percentiles are 0.

Overall: Comparing the mean and median helps identify skewness, while standard deviation and the interquartile range help understand the spread of the data. Fare has particularly high variability and positive skewness, while Age has a more moderate distribution.
